```python
import pandas as pd

# 1. Prepare unique symbols from options_df as a set
unique_symbols = set(options_df['symbol'].astype(str).str.strip().str.upper().unique())
total_initial = len(unique_symbols)

# 2. Prepare the read dataset nse_symbol values
read_symbols = set(
    read['nse_symbol'].astype(str).str.strip().str.upper().unique()
)

# 3. Match the options symbols against the read dataset's nse_symbol column
matched_symbols = unique_symbols.intersection(read_symbols)
missing_symbols = sorted(unique_symbols.difference(matched_symbols))

# 4. Report results for read dataset
print(f"Total unique symbols in options_df: {total_initial}")
print(f"Matches found in read['nse_symbol']: {len(matched_symbols)}")
print(f"Symbols not found in read['nse_symbol']: {len(missing_symbols)}")

# 5. Display sample matched and missing symbols
print("\nSample matched symbols:")
print(sorted(list(matched_symbols))[:50])

print("\nSample missing symbols:")
print(missing_symbols[:50])

# 6. Now check the unresolved missing symbols against three symboldata columns
symbol_map = symboldata.copy()
for col in ['nsesymbol', 'firstsymbol', 'latestsymbol']:
    symbol_map[col] = symbol_map[col].astype(str).str.strip().str.upper()

missing_df = pd.DataFrame({'symbol': missing_symbols})
missing_df['match_nse_symbol'] = missing_df['symbol'].isin(symbol_map['nsesymbol'])
missing_df['match_firstsymbol'] = missing_df['symbol'].isin(symbol_map['firstsymbol'])
missing_df['match_latestsymbol'] = missing_df['symbol'].isin(symbol_map['latestsymbol'])

missing_df['any_symboldata_match'] = (
    missing_df[['match_nse_symbol', 'match_firstsymbol', 'match_latestsymbol']].any(axis=1)
)

print("\nMissing symbols matching symboldata columns:")
print(missing_df['any_symboldata_match'].sum())
print("\nBreakdown by column:")
print(missing_df[['match_nse_symbol', 'match_firstsymbol', 'match_latestsymbol']].sum())

missing_df
```

```python
# Filter missing symbols that matched any symboldata column
matched_missing_df = missing_df[missing_df['any_symboldata_match']].copy()

symbol_map = symboldata.copy()
for col in ['nsesymbol', 'firstsymbol', 'latestsymbol']:
    symbol_map[col] = symbol_map[col].astype(str).str.strip().str.upper()

read_nse = set(read['nse_symbol'].astype(str).str.strip().str.upper().unique())

remaining = set(matched_missing_df['symbol'])
matched_order = []

for col in ['nsesymbol', 'firstsymbol', 'latestsymbol']:
    if not remaining:
        break

    candidates = symbol_map[symbol_map[col].isin(remaining)].copy()
    if candidates.empty:
        continue

    candidates['candidate_nse'] = candidates['nsesymbol'].astype(str).str.strip().str.upper()
    candidates = candidates[candidates['candidate_nse'].isin(read_nse)]
    if candidates.empty:
        continue

    for orig_sym, group in candidates.groupby(col):
        if orig_sym in remaining:
            matched_order.append({
                'symbol': orig_sym,
                'matched_on': col,
                'candidate_nse': group['candidate_nse'].iloc[0],
                'candidate_rows': len(group),
            })
            remaining.discard(orig_sym)

matched_order_df = pd.DataFrame(matched_order)

print("Matched missing symbols by symboldata columns:")
print(matched_order_df)
print(f"\nRemaining symbols after sequential matching: {len(remaining)}")
print(sorted(remaining)[:50])
```

```python
# Build the final two-column mapping dataset for direct and mediated matches
read_lookup = read.assign(
    nse_symbol=read['nse_symbol'].astype(str).str.strip().str.upper()
)[['nse_symbol', 'co_code']].drop_duplicates()

# Direct mappings from options_df symbols to read co_code
direct_matches = pd.DataFrame({'symbol': sorted(matched_symbols)})
direct_matches = direct_matches.merge(
    read_lookup,
    left_on='symbol',
    right_on='nse_symbol',
    how='left'
).drop(columns='nse_symbol')

# Mediated mappings via symbol_data -> candidate_nse -> read co_code
mediated_matches = matched_order_df.merge(
    read_lookup,
    left_on='candidate_nse',
    right_on='nse_symbol',
    how='left'
).rename(columns={'symbol': 'options_symbol'})
mediated_matches = mediated_matches[['options_symbol', 'co_code']].rename(columns={'options_symbol': 'symbol'})

# Final map, keeping only successful mappings and exact option symbols
options_to_co_code = pd.concat([direct_matches, mediated_matches], ignore_index=True)
options_to_co_code = options_to_co_code.dropna(subset=['co_code']).drop_duplicates().reset_index(drop=True)

print(f"Direct matches: {direct_matches['symbol'].nunique()}")
print(f"Mediated matches: {mediated_matches['symbol'].nunique()}")
print(f"Total mapped rows: {len(options_to_co_code)}")

options_to_co_code.head(50)
```

```python
# Report unmapped symbols from overall options_df
unique_options_symbols = set(
    options_df['symbol'].astype(str).str.strip().str.upper().unique()
)
mapped_symbols = set(
    options_to_co_code['symbol'].astype(str).str.strip().str.upper().unique()
)
missing_from_overall = sorted(unique_options_symbols.difference(mapped_symbols))

print(f"Total unique symbols in options_df: {len(unique_options_symbols)}")
print(f"Unique mapped symbols in options_to_co_code: {len(mapped_symbols)}")
print(f"Missing symbols count: {len(missing_from_overall)}")
print("\nFirst 35 missing symbols:")
print(missing_from_overall[:35])

missing_from_overall
```

```python
# Drop all rows whose symbol did not make it into the final mapped dataset
mapped_symbols = set(
    options_to_co_code['symbol'].astype(str).str.strip().str.upper().unique()
)

# Add co_code to options_df from the final symbol-to-co_code mapping
map_df = options_to_co_code[['symbol', 'co_code']].drop_duplicates().copy()
map_df['symbol'] = map_df['symbol'].astype(str).str.strip().str.upper()

options_df = options_df.copy()
options_df['symbol'] = options_df['symbol'].astype(str).str.strip().str.upper()

options_df = options_df.merge(
    map_df,
    on='symbol',
    how='left'
)

# Keep only rows whose symbol made it into the final mapped dataset
options_df = options_df[
    options_df['symbol'].isin(mapped_symbols)
].copy()

print(f"Rows remaining in options_df after cleanup: {len(options_df)}")
print(f"Unique symbols remaining: {options_df['symbol'].astype(str).str.strip().str.upper().nunique()}")
options_df.head()
```

```python
# Build a date-based lookup from closing_prices
closing_prices_lookup = closing_prices[['co_code', 'co_stkdate', 'nse_closing_price']].drop_duplicates().copy()
closing_prices_lookup['co_code'] = closing_prices_lookup['co_code'].astype('int64')
closing_prices_lookup['co_stkdate'] = pd.to_datetime(closing_prices_lookup['co_stkdate'], errors='coerce', dayfirst=True)

# Prepare options_df for joining
options_df = options_df.copy()
options_df['co_code'] = pd.to_numeric(options_df['co_code'], errors='coerce').astype('int64')
options_df['timestamp_dt'] = pd.to_datetime(options_df['timestamp'], errors='coerce', dayfirst=True)

# Merge on co_code + date and add the closing price column
options_df = options_df.merge(
    closing_prices_lookup.rename(columns={'co_stkdate': 'timestamp_dt'})[
        ['co_code', 'timestamp_dt', 'nse_closing_price']
    ],
    on=['co_code', 'timestamp_dt'],
    how='left',
    suffixes=('', '_closing_prices')
)

options_df['nse_closing_price'] = (
    options_df['nse_closing_price_closing_prices']
    .combine_first(options_df['nse_closing_price'])
)

options_df = options_df.drop(columns=['timestamp_dt', 'nse_closing_price_closing_prices'])

# Optional: check the result
options_df[['symbol', 'timestamp', 'co_code', 'nse_closing_price']].head()
```
